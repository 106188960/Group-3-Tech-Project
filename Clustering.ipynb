```python
import pandas as pd
import numpy as np
from sklearn import datasets
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
DATA_PATH = './output.csv.gz'
df = pd.read_csv(DATA_PATH)
df = df.drop(columns=["url", "domain"])
```

```python
X = df.drop('label', axis = 1)
y = df['label']

raw = KMeans(3, n_init=10, random_state=42).fit_predict(X)
X_scaled = StandardScaler().fit_transform(X)
scaled = KMeans(3, n_init=10, random_state=42).fit_predict(X_scaled)

print(np.bincount(raw))
print(np.bincount(scaled))
```

```python
# Create KMeans model and train
km = KMeans(n_clusters=15, n_init=5, random_state=45)  # TODO: fill in some numbers, does this sounds familiar?
labels = km.fit_predict(X_scaled)
print(km.inertia_, np.bincount(labels))
```

Codeismeasuringinertiaandsilhoure

```python
from sklearn.metrics import silhouette_score
for k in range(2, 9):
    km = KMeans(k, n_init=10, random_state=42)
    lab = km.fit_predict(X_scaled)
    print(k,
        round(km.inertia_),
        round(silhouette_score(X_scaled, lab), 3))
```

```python
df["cluster"] = labels
overall, spread = X.mean(), X.std()
for c in sorted(df.cluster.unique()):
    g = df[df.cluster == c]
    z = (g[X.columns].mean() - overall) / spread # standard deviations from normal
    high, low = z.nlargest(2), z.nsmallest(2)
    print(f'cluster {c} n={len(g)}')
    print(' more:', high.round(2).to_dict())
    print(' less:', low.round(2).to_dict())
```
